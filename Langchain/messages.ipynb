```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    model="openai/gpt-oss-120b",
    model_provider="groq",
    temperature=0.7
)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage
from pprint import PrettyPrinter
messages = [
SystemMessage("You are a poetry expert"),
HumanMessage("Write a poem on artificial intelligence")
]
response = model.invoke(messages)
print(response.content)
```

```python
ai_msg = AIMessage("I'd be happy to help you with that question!")
messages = [
SystemMessage("You are a helpful assistant"),
HumanMessage("Can you help me?"),
ai_msg, # Insert as if it came from the model
HumanMessage("Great! What's 2+2?"),
]
response = model.invoke(messages)
for msg in messages:
    print(msg.content)
print(response.content)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    HumanMessage("Say: I'd be happy to help you with that question!"),  # instruct model to output this
    HumanMessage("Great! What's 2+2?"),
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage

# Step 1: User asks
human_message = HumanMessage("What's the weather in San Francisco?")

# Step 2: Model tool call (simulate)
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_123"
    }]
)

# Step 3: Tool execution result
weather_result = "Sunny, 72F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

# Step 4: Pass full conversation to model
messages = [human_message, ai_message, tool_message]
response = model.invoke(messages)

print(response.content)
```

## Managing long message lists: trim_messages, filter_messages, merge_message_runs.

```python
from langchain_core.messages import (
    trim_messages,
    filter_messages,
    merge_message_runs,
    SystemMessage,
    HumanMessage,
    AIMessage,
)

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Hi"), AIMessage("Hello!"),
    HumanMessage("What's 2+2?"), AIMessage("4"),
    HumanMessage("What's the capital of France?"), AIMessage("Paris"),
    HumanMessage("And Germany?"),
]

# Trimmed messages
trimmed = trim_messages(messages, max_tokens=30, token_counter=len, strategy="last", include_system=True)
print(trimmed)

# Only human messages
humans_only = filter_messages(messages, include_types=["human"])
print(humans_only)

# Merge consecutive runs
merged = merge_message_runs([
    HumanMessage("Hi"),
    HumanMessage("Are you there?"),
    AIMessage("Yes!")
])
for msg in merged:
    print(f"{msg.type}: {msg.content}")
```
