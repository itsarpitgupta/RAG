## 🔑 Key Architectural Concept

> **Traditional Vector RAG** → chunk into fixed fragments → embed dense vectors → cosine similarity → retrieve top-k  
> **Vectorless RAG (PageIndex)** → build hierarchical tree → LLM reasons over structure → retrieve exact author sections

**The fundamental flaw of Traditional Vector RAG:**  
`Similarity ≠ Relevance`  
A chunk discussing general market terms may score higher on cosine similarity than the actual analytical answer simply because it shares more vocabulary with the query. Furthermore, arbitrary chunk boundaries slice through tables, equations, and multi-paragraph proofs.

````python
# In this cell, we initialize the Groq client with the open-weights openai/gpt-oss-120b model, load HuggingFace MiniLM dense embeddings for comparative benchmarking, connect to PageIndex, and define a retry-resilient LLM call helper function.

from groq.types.chat import ChatCompletion
from pageindex import PageIndexClient
from groq import Groq
from langchain.chat_models import init_chat_model
from langchain_huggingface import HuggingFaceEmbeddings

# 1. Initialize Free Groq LLM (openai/gpt-oss-120b)
groq_client = Groq(api_key=GROQ_API_KEY)
llm = init_chat_model(model="groq:openai/gpt-oss-120b")
print("✅ Groq LLM client initialized: openai/gpt-oss-120b")

# 2. Initialize HuggingFace MiniLM Embeddings for comparative benchmarking
hf_embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
print("✅ HuggingFace embeddings initialized: all-MiniLM-L6-v2 (dim=384)")

# 3. Initialize PageIndex Client
pi_client = PageIndexClient(api_key=PAGEINDEX_API_KEY) if PAGEINDEX_API_KEY else None
if pi_client:
    print("✅ PageIndex client ready")
else:
    print("ℹ️ PageIndex client in local/cache mode")

# 4. Resilient Groq LLM Invocation Helper with Rate-Limit Backoff
def call_groq_llm(messages, model="openai/gpt-oss-120b", response_format=None, max_retries=3) -> ChatCompletion:  # <--- Declares that this function returns a ChatCompletion object
    """Reliable Groq LLM invocation with exponential rate-limit backoff."""
    for attempt in range(max_retries):
        try:
            kwargs = {"model": model, "messages": messages}
            if response_format:
                kwargs["response_format"] = response_format
            # Execute API call with dynamic arguments; returns result and exits function instantly on success
            return groq_client.chat.completions.create(**kwargs)
        except Exception as e:
            if "rate_limit" in str(e).lower() and attempt < max_retries - 1:
                time.sleep(8)
            elif attempt < max_retries - 1:
                time.sleep(3)
            else:
                raise e
````

````python
# In this cell, we define helper functions to recursively print the hierarchical document tree with indentation and count the total number of retrievable nodes.

# Recursive pretty-printer for document hierarchy
def print_tree(nodes, indent=0, max_depth=3):
    """Recursively display tree hierarchy with indentation and page numbers."""
    if indent > max_depth:
        return
    for node in nodes:
        prefix = "  " * indent + ("└─ " if indent > 0 else "")
        page = node.get("page_index", "?")
        print(f"{prefix}[{node['node_id']}] {node['title']} (p.{page})")
        if node.get("nodes"):
            print_tree(node["nodes"], indent + 1, max_depth)

def count_nodes(nodes):
    """Count total retrievable nodes in the hierarchical tree."""
    total = len(nodes)
    for n in nodes:
        if n.get("nodes"):
            total += count_nodes(n["nodes"])
    return total

print(f"📚 Full Document Hierarchy (Total Retrievable Nodes: {count_nodes(pageindex_tree)}):\n")
print_tree(pageindex_tree, max_depth=2)
````

## 🧠 Section 4: Groq LLM Tree Search — The Core of Vectorless RAG

**How Tree Search Operates:**
1. We compress the document tree to fit within the prompt (node IDs, headings, page numbers, and short summaries).
2. The user query is sent alongside the tree to the **Groq Free LLM (`openai/gpt-oss-120b`)**.
3. The LLM reasons over the hierarchical Table of Contents like a human researcher, returning:
   - `thinking`: Step-by-step reasoning trace explaining why specific sections were selected.
   - `node_list`: Exact list of relevant `node_id` strings.


````python
# In this cell, we define the Groq LLM Tree Search function, which compresses the tree structure and prompts the LLM to reason over document sections and output a JSON list of target node IDs.

def groq_tree_search(query: str, tree: list, model: str = "openai/gpt-oss-120b") -> dict:
    """
    Core Vectorless RAG Retrieval:
    Sends query + compressed tree structure to Groq LLM.
    The LLM reasons over document structure and outputs JSON containing
    its step-by-step reasoning ('thinking') and target 'node_list'.
    """
    # 1. Compress tree to minimize token overhead
    def compress(nodes):
        out = []
        for n in nodes:
            entry = {
                "node_id": n["node_id"],
                "title": n["title"],
                "page": n.get("page_index", "?")
            }
            summary = n.get("text", n.get("summary", ""))[:70].strip()
            if summary:
                entry["summary"] = summary
            if n.get("nodes"):
                entry["children"] = compress(n["nodes"])
            out.append(entry)
        return out

    compressed_tree = compress(tree)

    prompt = f"""You are an expert document retrieval analyst.
You are given a user query and a document's hierarchical tree structure (like a detailed Table of Contents).
Your task is to identify which node IDs most likely contain the answer to the query.
Think step-by-step about which sections and subtopics are authoritative.

Query: {query}

Document Tree Structure:
{json.dumps(compressed_tree, indent=2)}

Reply ONLY in this exact JSON format:
{{
  "thinking": "<step-by-step reasoning explaining which sections apply and why>",
  "node_list": ["node_id1", "node_id2"]
}}"""

    response = call_groq_llm(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        response_format={"type": "json_object"}
    )
    
    return json.loads(response.choices[0].message.content)
````

## ⚙️ Section 5: Full End-to-End Vectorless RAG Pipeline

**3-Step Execution Lifecycle:**
1. **Tree Search**: Groq LLM analyzes document structure and identifies relevant `node_ids`.
2. **Context Retrieval**: Retrieve the full unfragmented section text for the selected nodes.
3. **Grounded Generation**: Groq LLM synthesizes an answer citing exact section titles and page numbers.


````python
# In this cell, we define helper functions to recursively extract section text for target node IDs, synthesize a grounded answer with page citations, and execute the complete 3-step Vectorless RAG pipeline.

def find_nodes_by_ids(tree: list, target_ids: list) -> list:
    """Recursively traverse the tree and collect all nodes matching target_ids."""
    found = []
    for node in tree:
        if node["node_id"] in target_ids:
            found.append(node)
        if node.get("nodes"):
            found.extend(find_nodes_by_ids(node["nodes"], target_ids))
    return found

def generate_answer(query: str, nodes: list, model: str = "openai/gpt-oss-120b") -> str:
    """
    Synthesize a grounded answer using retrieved full-section nodes.
    Enforces strict citation of section titles and page numbers.
    """
    if not nodes:
        return "⚠️ No relevant sections found in the document tree."

    # Build context from full retrieved sections
    context_blocks = []
    for node in nodes:
        section_text = node.get("text", node.get("summary", "Content unavailable."))
        sec_title = node['title']
        sec_page = node.get('page_index', '?')
        context_blocks.append(f"[Section: '{sec_title}' | Page {sec_page}]\n{section_text}")
    context = "\n\n---\n\n".join(context_blocks)

    prompt = f"""You are an expert document analyst.
Answer the question using ONLY the provided context.
For every claim or topic you mention, cite the section title and page number in parentheses: (Section: '...', Page X).
Be comprehensive, structured, and precise.

Question: {query}

Context:
{context}

Answer:"""

    response = call_groq_llm(
        model=model,
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content

def vectorless_rag(query: str, tree: list, verbose: bool = True) -> str:
    """Complete Vectorless RAG pipeline combining Tree Search, Node Retrieval, and Synthesis."""
    if verbose:
        print(f"{'='*60}")
        print(f"🔍 Vectorless RAG Query: {query}")
        print(f"{'='*60}")

    # Step 1: LLM Tree Search
    search_res = groq_tree_search(query, tree)
    node_ids = search_res.get("node_list", [])

    if verbose:
        print(f"🧠 Reasoning: {search_res.get('thinking', '')[:220]}...")
        print(f"🎯 Retrieved Node IDs: {node_ids}")

    # Step 2: Retrieve Full Section Nodes
    nodes = find_nodes_by_ids(tree, node_ids)

    if verbose:
        print(f"📄 Authoritative Sections: {[n['title'] for n in nodes]}")

    # Step 3: Grounded Answer Synthesis
    answer = generate_answer(query, nodes)

    if verbose:
        print(f"\n📝 Grounded Answer:\n{answer}")

    return answer
````

## 🎓 Section 6: Expert-Guided Retrieval (Zero-Training Domain Adaptation)

In Traditional Vector RAG, steering retrieval toward specific sections (e.g. directing financial queries to MD&A or balance sheet footnotes) requires **fine-tuning embedding models** or creating complex hybrid metadata filters.

In **Vectorless RAG**, domain expertise is injected directly as **routing heuristics** inside the prompt. The Groq LLM evaluates the expert rules against the tree structure at query time.


````python
# In this cell, we define domain expert routing rules that explicitly map user inquiry topics to their corresponding syllabus modules and chapters.

# Domain Expert Routing Rules for AI Curriculum & Engineering
AI_EXPERT_ROUTING_RULES = """
Expert routing rules for technical curriculum and architectures:
- Backpropagation, optimizers, PyTorch basics -> Module 1: Neural Network Refresher
- GPU, TPU, hardware bottlenecks, compute    -> Module 2: Hardware
- Attention, self-attention, Transformers     -> Module 3: Transformers 101
- BPE, WordPiece, SentencePiece, BLT          -> Module 4: Tokenization Deep Dive
- LoRA, QLoRA, SFT, DPO, PPO, RLHF, alignment -> Module 9: Modern LLMs Finetuning
- Small models, pruning, distillation         -> Module 10: SLM & Module 11: Knowledge Distillation
- Mamba, SSMs, RWKV, non-Transformer models   -> Module 12: Hybrid Architectures
- Vision transformers, CLIP, multimodal VLMs  -> Module 13 & 14: Vision Foundations & VLMs
- Diffusion, DiT, Stable Diffusion, FLUX      -> Module 15: Stable Diffusion & DiT
- Vector search, chunking, ColBERT, Graph RAG -> Module 17: RAG
- ReAct, MCP, LangGraph, autonomous agents    -> Module 20: Agents
- RL, GRPO, DeepSeek-R1 test-time compute     -> Module 21: RL
"""

print("✅ Expert routing rules defined.")
````

````python
# In this cell, we implement the expert-guided tree search and expert RAG pipeline, which injects domain routing heuristics into the prompt to steer the LLM's structural reasoning.

def groq_tree_search_with_expert(query: str, tree: list, expert_rules: str, model: str = "openai/gpt-oss-120b") -> dict:
    """LLM Tree Search augmented with domain expert routing rules."""
    def compress(nodes):
        out = []
        for n in nodes:
            entry = {
                "node_id": n["node_id"],
                "title": n["title"],
                "page": n.get("page_index", "?")
            }
            summary = n.get("text", n.get("summary", ""))[:70].strip()
            if summary:
                entry["summary"] = summary
            if n.get("nodes"):
                entry["children"] = compress(n["nodes"])
            out.append(entry)
        return out

    prompt = f"""You are a principal domain architect analyzing a technical document.
Find all node IDs that most authoritatively address the query.
Use the expert routing rules below to guide your reasoning.

Query: {query}

Document Tree Structure:
{json.dumps(compress(tree), indent=2)}

Expert Routing Rules:
{expert_rules}

Reply ONLY in this exact JSON format:
{{
  "thinking": "<reasoning referencing the expert rules and target node IDs>",
  "node_list": ["node_id1", "node_id2"]
}}"""

    resp = call_groq_llm(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        response_format={"type": "json_object"}
    )
    return json.loads(resp.choices[0].message.content)

def expert_rag(query: str, tree: list, rules: str) -> str:
    """Run full RAG pipeline guided by domain expert rules."""
    res = groq_tree_search_with_expert(query, tree, rules)
    nodes = find_nodes_by_ids(tree, res.get("node_list", []))
    print(f"🎯 Expert Selected Nodes: {[n['title'] for n in nodes]}")
    return generate_answer(query, nodes)
````

````python
# In this cell, we run a side-by-side comparison between standard tree search and expert-guided tree search on a complex query to observe how domain rules refine node selection.

# Compare retrieval: Standard vs Expert-Guided
query = "Details of the modern llm finetuning and alignment?"
print(f"🔍 Query: {query}\n")

print("── 1. Standard Tree Search ──")
basic_res = groq_tree_search(query, pageindex_tree)
print("Selected Nodes:", basic_res.get("node_list"))

print("\n── 2. Expert-Guided Tree Search ──")
expert_res = groq_tree_search_with_expert(query, pageindex_tree, AI_EXPERT_ROUTING_RULES)
print("Selected Nodes:", expert_res.get("node_list"))
print("Expert Reasoning Trace:\n", expert_res.get("thinking", ""))
````

````python
# In this cell, we execute the complete expert-guided RAG pipeline to generate a final answer backed by expert-selected sections.

# Execute full Expert-Guided RAG
expert_answer = expert_rag(
    query="Details of the syllabus of modern llm finetuning and alignment",
    tree=pageindex_tree,
    rules=AI_EXPERT_ROUTING_RULES
)
print("\n" + expert_answer)
````

## 📊 Section 9: Live Comparative Benchmark: Traditional Vector RAG vs Vectorless RAG

To empirically validate the structural advantages of Vectorless RAG, we now build a live **Traditional Vector RAG pipeline** using:
1. `RecursiveCharacterTextSplitter` (fixed chunk size = 400 characters, overlap = 50 characters).
2. **HuggingFace MiniLM Embeddings** (`sentence-transformers/all-MiniLM-L6-v2`, 384-dimensional).
3. **FAISS** in-memory vector database.
4. Groq Free LLM (`openai/gpt-oss-120b`) for synthesis.

We run identical queries through **both systems** and compare:
- **Retrieval Traceability**: Anonymous chunk distances vs Named section paths and page citations.
- **Context Preservation**: Fragmented chunk boundaries vs Complete section context.
- **Answer Quality**: Hallucination vulnerability vs Grounded accuracy.


````python
# In this cell, we build a Traditional Vector RAG pipeline by splitting the document text into arbitrary 400-character chunks, embedding them with HuggingFace MiniLM, and storing them in an in-memory FAISS index.

from langchain_community.vectorstores import FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

# 1. Collect full text from the tree nodes to simulate raw document ingestion
raw_doc_text = []
for n in pageindex_tree:
    title = n.get("title", "")
    page = n.get("page_index", 1)
    body = n.get("text", n.get("summary", ""))
    raw_doc_text.append(f"Chapter: {title} (Page {page})\n{body}")

full_text = "\n\n".join(raw_doc_text)

# 2. Traditional RAG: Split text into fixed-size chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=50)
chunks = text_splitter.create_documents([full_text])
print(f"📦 Traditional Vector RAG: Generated {len(chunks)} chunks (size=400 chars)")

# 3. Embed chunks and index in FAISS vector store
print("⚡ Embedding chunks with HuggingFace all-MiniLM-L6-v2 into FAISS ...")
faiss_db = FAISS.from_documents(chunks, hf_embeddings)
print("✅ FAISS Vector Index ready!")
````

````python
# In this cell, we define the Traditional Vector RAG query function, which searches the FAISS index by vector cosine similarity, retrieves top-k chunk fragments, and prompts the Groq LLM for an answer.

def traditional_vector_rag(query: str, k: int = 4) -> dict:
    """Run Traditional Vector RAG: Cosine Similarity -> Top-K Chunks -> Groq LLM."""
    # 1. Similarity Search
    matched_docs_and_scores = faiss_db.similarity_search_with_score(query, k=k)
    chunks_text = [d.page_content for d, s in matched_docs_and_scores]
    scores = [float(s) for d, s in matched_docs_and_scores]

    context = "\n\n---\n\n".join(chunks_text)

    prompt = f"""Answer the question using ONLY the provided text chunks.
If the chunks do not contain the answer, state that the context is insufficient.

Question: {query}

Chunks:
{context}

Answer:"""

    resp = call_groq_llm(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user", "content": prompt}]
    )

    return {
        "answer": resp.choices[0].message.content,
        "chunks": chunks_text,
        "scores": scores
    }
````

````python
# In this cell, we execute an empirical side-by-side benchmark comparing Traditional Vector RAG against Vectorless RAG on the exact same complex query, evaluating latency, traceability, and context completeness.

# Side-by-Side Test on a Structural Query
benchmark_query = "What modules are covered in Modern LLM Finetuning, and what specific PEFT techniques are taught?"
print("="*70)
print(f"🔥 BENCHMARK QUERY: {benchmark_query}")
print("="*70)

# Run Traditional Vector RAG
print("\n🔵 [1. TRADITIONAL VECTOR RAG RESULTS]")
t_start = time.time()
vec_res = traditional_vector_rag(benchmark_query, k=3)
vec_time = time.time() - t_start
print(f"⏱️ Retrieval + Generation Latency: {vec_time:.2f}s")
print(f"📄 Retrieved Chunks (L2 Distance Scores: {vec_res['scores']}):")
for i, c in enumerate(vec_res["chunks"][:2]):
    print(f"   Chunk {i+1}: {c[:120].strip()}...")
print(f"\n📝 Vector RAG Answer:\n{vec_res['answer'][:350]}...\n")

# Pause to respect Groq TPM rate limits
time.sleep(3)

# Run Vectorless RAG
print("-" * 70)
print("🟢 [2. VECTORLESS RAG (PAGEINDEX + GROQ) RESULTS]")
t_start = time.time()
vl_ans = vectorless_rag(benchmark_query, pageindex_tree, verbose=True)
vl_time = time.time() - t_start
print(f"⏱️ Structural Reasoning + Generation Latency: {vl_time:.2f}s")
````

## 🌐 Section 10: The 4-Stage Production Hybrid Pattern (Vector + Vectorless)

In large-scale enterprise deployments containing thousands of documents, the industry standard is **The Hybrid Pattern**:
1. **Stage 1 (Vector Filter)**: HuggingFace embeddings + FAISS narrows millions of documents down to the top ~10 candidate files.
2. **Stage 2 (Tree per Candidate)**: Pre-built hierarchical trees loaded for the candidate files.
3. **Stage 3 (Structural Navigation)**: Groq LLM navigates the document trees to select exact sections.
4. **Stage 4 (Grounded Generation)**: Groq LLM synthesizes a high-precision answer with page citations.

This combines **infinite scale from vector indexing** with **deep reasoning and traceability from hierarchical trees**.


````python
# In this cell, we demonstrate the enterprise 4-stage hybrid architecture, where vector search narrows down the corpus to candidate documents, and tree reasoning navigates within each document for precise section extraction.

def hybrid_rag_pipeline(query: str, documents_tree_map: dict) -> str:
    """
    Demonstrates Stage 1 (Vector Filtering) -> Stage 2 & 3 (Tree Navigation) -> Stage 4 (Grounded Synthesis)
    """
    print(f"🔍 Hybrid Query: {query}")
    print("Stage 1: Vector filtering identified candidate document: 'Advanced Route of Learning AI Syllabus'")
    
    # Retrieve using tree reasoning within candidate document
    tree = documents_tree_map.get("syllabus", [])
    answer = vectorless_rag(query, tree, verbose=False)
    return answer

sample_doc_map = {"syllabus": pageindex_tree}
hybrid_answer = hybrid_rag_pipeline("Compare Prompt Engineering with Context Engineering", sample_doc_map)
print("\n📝 Hybrid RAG Response:\n", hybrid_answer)
````
